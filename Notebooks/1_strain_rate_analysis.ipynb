## 8. Strain rates due to elastic heterogeneity and viscoelastic earthquake cycle effects

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat, savemat
from scipy.interpolate import interp1d

# Load .mat files
layered_data = loadmat('../Input_data/datasets/threshold_curve/layered_residual.mat')
visco_data = loadmat('../Input_data/datasets/threshold_curve/visco_residual.mat')
lateral_data = loadmat('../Input_data/datasets/threshold_curve/lateral_residual.mat')

# Extract variables 
scale = 1e-9
distances_layered = layered_data['distances_layered'].squeeze()
envelope_layered = layered_data['envelope_layered'].squeeze()
envelope_layered /= scale

distances_visco = visco_data['distances_visco'].squeeze()
envelope_visco = visco_data['envelope_visco'].squeeze()
envelope_visco /= scale

distances_lateral = lateral_data['distances_lateral'].squeeze()
envelope_lateral = lateral_data['envelope_lateral'].squeeze()
envelope_lateral /= scale

# Append final point for layered model
distances_layered = np.append(distances_layered, 500)

envelope_layered = np.append(envelope_layered, 0)

# Interpolation grid
distances = np.linspace(0, 360, 360)

# Interpolate
interp_visco = interp1d(distances_visco, envelope_visco,
                        bounds_error=False, fill_value=np.nan)
interp_lateral = interp1d(distances_lateral, envelope_lateral,
                          bounds_error=False, fill_value=np.nan)
interp_layered = interp1d(distances_layered, envelope_layered,
                          bounds_error=False, fill_value=np.nan)

envelope_visco = interp_visco(distances)
envelope_lateral = interp_lateral(distances)
envelope_layered = interp_layered(distances)

# Plot
plt.figure()
plt.plot(distances, envelope_visco, linewidth=2, color=(26/255, 143/255, 227/255), label='Viscoelastic mantle flow')
plt.plot(distances, envelope_layered, linewidth=2, color=(252/255, 163/255, 17/255), label='Vertical heterogeneity')
plt.plot(distances, envelope_lateral, linewidth=2, color=(209/255, 17/255, 73/255), label='Lateral heterogeneity')

# Replace NaNs with 0
envelope_visco = np.nan_to_num(envelope_visco)
envelope_lateral = np.nan_to_num(envelope_lateral)
envelope_layered = np.nan_to_num(envelope_layered)

# Total curve
total = envelope_visco + envelope_lateral + envelope_layered
plt.plot(distances, total, linewidth=2, color='black', label='Total contribution')

# Formatting
plt.legend(fontsize=14)
plt.xlabel('Distance from fault (km)', fontsize=14)
plt.ylabel(r'Max. shear strain-rate per unit slip-rate', fontsize=14)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

ax = plt.gca()
ax.text(
    0.58, 0.13,
    r'$10^{-9}$ yr$^{-1}$ per 1 mm yr$^{-1}$',
    transform=ax.transAxes,
    fontsize=12,
    va='top',
    ha='left'
)

# set y-axis limit to 4.8
plt.ylim(0, 4.8)

# Save figure
plt.savefig("../output_data/reproducible_research_with_bounds/figures/threshold_curve_decomposition.pdf", dpi=600)

plt.show()

```

## 9. Relative contribution of off-fault deformation to the total strain rate field

```python
# Load on-fault and off-fault strain rate data
on_fault_file = "../Input_data/datasets/CAL_strain_rates/strain_rates_CAL_backslip.csv"
off_fault_file = "../Input_data/datasets/CAL_strain_rates/strain_rates_CAL_moment_souces.csv"

on_fault_df = pd.read_csv(
            on_fault_file,
            sep=r"\s+",
            header=None,
            names=['lon','lat','mean_maxshear','std_maxshear','mean_dilatation',
                   'std_dilatation','Exx_mean','Exy_mean','Eyy_mean',
                   'Exx_std','Exy_std','Eyy_std']
        )

off_fault_df = pd.read_csv(
            off_fault_file,
            sep=r"\s+",
            header=None,
            names=['lon','lat','mean_maxshear','std_maxshear','mean_dilatation',
                   'std_dilatation','Exx_mean','Exy_mean','Eyy_mean',
                   'Exx_std','Exy_std','Eyy_std']
        )

# merge on lon and lat
merged_df = pd.merge(on_fault_df, off_fault_df, on=['lon', 'lat'], suffixes=('_on_fault', '_off_fault'))

# Compute the total strain rate by summing on-fault and off-fault contributions and then compute the % contribution of off-fault
merged_df['total_maxshear'] = merged_df['mean_maxshear_on_fault'] + merged_df['mean_maxshear_off_fault']
merged_df['off_fault_contribution_percent'] = (merged_df['mean_maxshear_off_fault'] / merged_df['total_maxshear']) * 100

# Drop NaNs 
merged_df = merged_df.dropna(subset=["off_fault_contribution_percent"])

# Compute Frobenius norm of the strain rate tensor for on-fault and off-fault
merged_df['on_fault_frobenius'] = np.sqrt(merged_df['Exx_mean_on_fault']**2 + 2*merged_df['Exy_mean_on_fault']**2 + merged_df['Eyy_mean_on_fault']**2)
merged_df['off_fault_frobenius'] = np.sqrt(merged_df['Exx_mean_off_fault']**2 + 2*merged_df['Exy_mean_off_fault']**2 + merged_df['Eyy_mean_off_fault']**2)

# Convert from microstrain/yr to nanostrain/yr
merged_df['on_fault_frobenius'] *= 1000
merged_df['off_fault_frobenius'] *= 1000

# Remove lines where Frobenius norm of both on-fault and off-fault is below 10 nanostrain/yr
threshold = 10  # nanostrain/yr
merged_df = merged_df[(merged_df['on_fault_frobenius'] >= threshold) & (merged_df['off_fault_frobenius'] >= threshold)] 

# reset index
merged_df = merged_df.reset_index(drop=True)
```

```python
import pandas as pd

# load file with strain rate contributions from elastic heterogeneity and mantle flow on the same observation points as our estimated on-fault and off-fault strain rates
thresholds_df = pd.read_csv("../Input_data/datasets/threshold_curve/threshold_model_on_obs_points.csv", header=None)
thresholds_df.columns = ['lon', 'lat', 'threshold_max_shear']

thresholds_df['lon'] = thresholds_df['lon'].round(2)
thresholds_df['lat'] = thresholds_df['lat'].round(3)

# Combine the merged_df with the thresholds_df based on lon and lat
combined_df_model_and_threshold = pd.merge(merged_df, thresholds_df, on=['lon', 'lat'], how='inner')

# Convert mean_maxshear_off_fault and threshold_max_shear to the same units (nanostrain/yr)
combined_df_model_and_threshold["mean_maxshear_off_fault_nstrain"] = (
    combined_df_model_and_threshold["mean_maxshear_off_fault"] * 1000
)
combined_df_model_and_threshold['threshold_max_shear'] *= 10**9  # Convert from strain/yr to nanostrain/yr

# Remove rows where mean_maxshear_off_fault is below the threshold value from the thresholds_df
combined_df_model_and_threshold = combined_df_model_and_threshold[combined_df_model_and_threshold['mean_maxshear_off_fault_nstrain'] >= combined_df_model_and_threshold['threshold_max_shear']]

# reset index
combined_df_model_and_threshold = combined_df_model_and_threshold.reset_index(drop=True)

# compute sum of mean_maxshear_off_fault and mean_maxshear_on_fault 
merged_df['total_maxshear'] = (
    merged_df['mean_maxshear_on_fault'] +
    merged_df['mean_maxshear_off_fault']
)

ratio_statewide = 100 * (
    merged_df['mean_maxshear_off_fault'].sum() /
    merged_df['total_maxshear'].sum()
)

print("Statewide off-fault %:", ratio_statewide)
```
